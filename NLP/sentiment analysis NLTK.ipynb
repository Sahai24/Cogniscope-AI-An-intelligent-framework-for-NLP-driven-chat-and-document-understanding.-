```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
#nltk.download('vader_lexicon')
```

```python
sia = SentimentIntensityAnalyzer()
```

```python
def analyze_sentiment_nltk(text):
    sentiment_score = sia.polarity_scores(text)
    polarity = sentiment_score['compound']
    
    if polarity < -0.2:
        sentiment = "Negative"
    elif polarity > 0.2:
        sentiment = "Positive"
    else:
        sentiment = "Neutral"
    
    return {"Sentiment": sentiment, "Polarity": polarity}
```

```python
texts = [
   "This laptop is fantastic! Super fast performance and excellent battery life.",  # Positive
    "I had a wonderful time at the resort. The staff was friendly and the view was breathtaking!",  # Positive
    "The smartphone exceeded my expectations. Great camera, smooth UI, and solid build quality!",  # Positive
    "Extremely disappointed with the quality of this product. It broke within two days of use!",  # Negative
    "The support team was unhelpful and kept redirecting me without resolving my issue.",  # Negative
    "I would never buy from this brand again. The worst customer experience ever!",  # Negative
    "The packaging was decent, but I haven’t tried the product yet.",  # Neutral
    "It's an okay book, not the best I’ve read, but not the worst either.",  # Neutral
    "Delivery was on time, but I haven’t checked if the product works properly.",  # Neutral
    "The service was average, nothing too bad but nothing extraordinary either."  # Neutral
]
```

```python
for i, text in enumerate(texts, 1):
    result = analyze_sentiment_nltk(text)
    print(f"Text {i}: {text}\nResult: {result}\n")
```
