```python
import spacy
from collections import Counter
import heapq
from docx import Document
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def read_file(file_path):

    if file_path.endswith(".txt"):
        with open(file_path, "r", encoding="utf-8") as file:
            return file.read()
    elif file_path.endswith(".docx"):
        doc = Document(file_path)
        return "\n".join([para.text for para in doc.paragraphs])
    else:
        raise ValueError("Unsupported file format. Please use .txt or .docx")
```

```python
def summarize_text(text, summary_ratio=0.2):

    doc = nlp(text)

    word_frequencies = Counter()
    for word in doc:
        if not word.is_stop and not word.is_punct:
            word_frequencies[word.text.lower()] += 1

    max_freq = max(word_frequencies.values(), default=1)
    for word in word_frequencies:
        word_frequencies[word] /= max_freq

    sentence_scores = {}
    sentences = list(doc.sents)

    for sent in sentences:
        for word in sent:
            if word.text.lower() in word_frequencies:
                sentence_scores[sent] = sentence_scores.get(sent, 0) + word_frequencies[word.text.lower()]

    num_sentences = max(1, int(len(sentences) * summary_ratio))
    top_sentences = heapq.nlargest(num_sentences, sentence_scores, key=sentence_scores.get)

    return " ".join([sent.text for sent in sorted(top_sentences, key=lambda s: s.start)])
```

```python
file_path = "Elon Musk.docx" 
text = read_file(file_path)
```

```python
summary = summarize_text(text, summary_ratio=0.1)
print("Summary:\n", summary)
```
